```python
import numpy as np
scores=np.array([45,78,56,89,90,67,34,23,88,76,54,61,72,48,59,92,81,39,64,70])
average=np.mean(scores)
highest=np.max(scores)
lowest=np.min(scores)
print("Average Score:",average)
print("Lowest Score:",highest)
print("Lowest Score:",lowest)
pass_mark=50
print("\nPass/Fail Status")
for i,score in enumerate(scores):
  if score >=pass_mark:
    print(f"Student {i+1}: Pass")
  else:
    print(f"Student {i+1}:Fail")
print("\nGrades")
for i, s in enumerate(scores):
  if s >=90:
    grade="A"
  elif s >=75:
    grade="B"
  elif s >=60:
    grade="C"
  elif s >=50:
    grade="D"
  else:
    grade = "F"
  print("Student",i+1,"Grade:",grade)
```

```python
import numpy as np
steps=np.array([5000,6200,7000,8000,9000,6500,7200,8100,9500,4000])
print("Total steps:",steps.sum())
print("Average steps:",steps.mean())
print("Maximum steps:",steps.max())
print("\nHigh activity days(>8000):")
for i in range(len(steps)):
  if steps[i] > 8000:
    print("Day",i+1,":",steps[i])
```

```python
import numpy as np

# Marks of students in Math, Science, English
marks = np.array([
[78,85,90],
[65,70,72],
[88,92,95],
[55,60,58],
[80,75,82]
])

subjects = ["Math","Science","English"]

# Average
avg = marks.mean(axis=0)

# Highest
high = marks.max(axis=0)

# Lowest
low = marks.min(axis=0)

for i in range(3):
    print(subjects[i])
    print("Average:",avg[i])
    print("Highest:",high[i])
    print("Lowest:",low[i])
    print()
```

```python
import numpy as np

Z1 = 10 + 5j
Z2 = 8 - 3j
Z3 = 6 + 2j


Z_series = Z1 + Z2 + Z3


Z_parallel = 1 / (1/Z1 + 1/Z2 + 1/Z3)

print("Impedance Z1 =", Z1)
print("Impedance Z2 =", Z2)
print("Impedance Z3 =", Z3)

print("Total Series Impedance =", Z_series)
print("Total Parallel Impedance =", Z_parallel)
```

```python
import numpy as np


stock1 = np.array([
    [100,110,95],
    [102,112,97],
    [105,115,100],
    [108,118,102],
    [110,120,105]
])
stock2 = np.array([
    [98,108,94],
    [101,111,96],
    [103,113,98],
    [107,117,101],
    [109,119,104]
])


sum_matrix = stock1 + stock2


diff_matrix = stock1 - stock2

print("Stock 1 Data:\n",stock1)
print("Stock 2 Data:\n",stock2)

print("Element-wise Sum:\n",sum_matrix)
print("Element-wise Difference:\n",diff_matrix)
```

```python
import pandas as pd

data = {
'Product':['Laptop','Smartphone','Tablet','Monitor','Keyboard'],
'Price':[1200,800,400,300,50],
'Quantity':[10,25,15,8,40]
}

df = pd.DataFrame(data)

print("Inventory Data:\n",df)

# Total value of each product
df["Total_Value"] = df["Price"] * df["Quantity"]

print("\nUpdated Inventory:\n",df)

print("\nTotal Inventory Value =", df["Total_Value"].sum())
```
