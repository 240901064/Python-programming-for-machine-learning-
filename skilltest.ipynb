```python
import numpy as np
scores=np.array([45,78,56,89,90,67,34,23,88,76,54,61,72,48,59,92,81,39,64,70])
average=np.mean(scores)
highest=np.max(scores)
lowest=np.min(scores)
print("Average Score:",average)
print("Lowest Score:",highest)
print("Lowest Score:",lowest)
pass_mark=50
print("\nPass/Fail Status")
for i,score in enumerate(scores):
  if score >=pass_mark:
    print(f"Student {i+1}: Pass")
  else:
    print(f"Student {i+1}:Fail")
print("\nGrades")
for i, s in enumerate(scores):
  if s >=90:
    grade="A"
  elif s >=75:
    grade="B"
  elif s >=60:
    grade="C"
  elif s >=50:
    grade="D"
  else:
    grade = "F"
  print("Student",i+1,"Grade:",grade)
```

```python
import numpy as np
steps=np.array([5000,6200,7000,8000,9000,6500,7200,8100,9500,4000])
print("Total steps:",steps.sum())
print("Average steps:",steps.mean())
print("Maximum steps:",steps.max())
print("\nHigh activity days(>8000):")
for i in range(len(steps)):
  if steps[i] > 8000:
    print("Day",i+1,":",steps[i])
```
